## I have to reuse same function for team members and manager ,but that have small  Difference.how can I Reuse that

* Both Team member and Manager should able to login,But:
* **Team member login** - View task and sumbit task.
* **manager login** - View task and assign Task.

## Without Oops

```python
def login(user):
    return f"{user['name']} has loggeed in."

def view_task(user):
    return f"user{['name']} has view assigned Task."

def sumbit_task(user):
    if user ['role'] == 'Team Member':
        return f"{user['name']} has sumbitted the Task."
    else:
        return "Team members only sumbit the Task."

def assign_task(user):
    if user ['role'] == 'Manager':
        return f"{user['name']} has assigned a new Task."
    else:
        return "Manager only can assign Task."


#sample user:
john = {'name':'john','role':'Team Member'}
nasir ={'name':'nasir','role':'Manager'}

#function call:
print(login(john))
print(view_task(john))
print(sumbit_task(john))
print(assign_task(john))
print("====================")
print(login(nasir))
print(view_task(nasir))
print(sumbit_task(nasir))
print(assign_task(nasir))

    
```

### Drawbacks Without OOPS:
#### Limitation                                          Explanation

**No Reusability of Logic**                  *You can't bundle shared behavior. You repeat logic checks (if role ...) again and again.*

**X Hard to Extend**                         Adding a new role (like Admin) means changing all your functions to include new if conditions.

**X No Structure**                           No way to logically group related functionality. Tasks for team members and managers are spread across 
                                             multiple function

**X No Data Binding**                        Data (name, role) is not attached to behavior. You have to keep passing the user dict again and again.

**X No Inheritance**                         You can't inherit shared behavior. Even common methods like login() get repeated logic.


## With Oops

**Real World Anology 1**

```python
class Employee:
    def __init__(self, name):
        self.name = name 
        
    def login(self):
       return f" {self.name} has logged in."
    
    def view_tasks(self):
       return f" {self.name} can view assigned tasks!"

class TeamMember(Employee):
    def submit_task(self):
       return f"{self.name} has submitted the task."

class Manager(Employee):
    def assign_task(self):
       return f"{self.name} has assigned a new task."
       
# Usage
john  = TeamMember("John")
alice = Manager("Nasir")

print(john.login())
print(john.submit_task())
print(alice.login())
print(alice.assign_task())
```

### With oops the code clean,scalable,Structured.

## I want to hide Sensitive data so other can't access or change data in future 

**Withot Oops**

```python
#No way to hide data here
user_data = {"password":"1234"}

def get_password():
    return user_data["password"]

get_password()
```

#### with Oops

```python
class secure_user:
    def  __init__(self,password):
        self._password = password

    def check_password(self,attempt):
        return self._password == attempt

user = secure_user("1234")
print(user.check_password("1234"))
print(user._password)
```

## Benefit : (Encapsulation/Privacy)
